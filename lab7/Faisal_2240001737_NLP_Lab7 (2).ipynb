# Part A — Decoder-Only Model

## Decoder-only example: GPT-style Text Generation

A decoder-only model predicts the next token based on previous tokens.

We will use **DistilGPT-2**, a small GPT-style model.


````python
generator = pipeline(
    "text-generation",
    model="distilgpt2",
    framework="pt",
    device=-1,
    model_kwargs={"use_safetensors": True}
)

prompt = "Natural language processing is"
output = generator(
    prompt,
    max_new_tokens=30,
    do_sample=False,
    pad_token_id=generator.tokenizer.eos_token_id
)
print(output[0]["generated_text"])
````

# Part B — Encoder–Decoder Model

## Encoder–decoder example: Instruction / Text-to-Text Generation

Encoder–decoder models contain two main parts:

1. The **encoder** reads the input.
2. The **decoder** produces the output.

We will use **FLAN-T5-small**.


````python
tokenizer = AutoTokenizer.from_pretrained("google/flan-t5-small")
model = AutoModelForSeq2SeqLM.from_pretrained(
    "google/flan-t5-small", use_safetensors=True
)
model.eval()

prompt = "Translate to French: I love natural language processing."
inputs = tokenizer(prompt, return_tensors="pt")
with torch.inference_mode():
    output = model.generate(**inputs, max_new_tokens=60, do_sample=False)

print(tokenizer.decode(output[0], skip_special_tokens=True))
````

# Part C — Encoder-Only Model

## Encoder-only example: Sentiment Classification

Encoder-only models read the full input and create a representation of it.

They are very useful when the task is to **understand or classify text**.

Here we use a DistilBERT sentiment model.


````python
classifier = pipeline(
    "sentiment-analysis",
    model="distilbert-base-uncased-finetuned-sst-2-english",
    framework="pt",
    device=-1,
    model_kwargs={"use_safetensors": True}
)

text = "This NLP lab is easy and interesting."
result = classifier(text)
print(result)
````

## Task 3 — Encoder–Decoder Model

Use the FLAN-T5 model to perform the following:

- Summarize a short paragraph.


````python
paragraph = (
    "The university library has extended its opening hours during exam week. "
    "It now stays open until midnight so students have more time to study. "
    "Extra quiet rooms are available, and students can reserve group study rooms online. "
    "The library also provides free workshops to help students prepare for their exams."
)

summary_prompt = "Summarize the following paragraph: " + paragraph
inputs = tokenizer(
    summary_prompt, return_tensors="pt", truncation=True, max_length=512
)
with torch.inference_mode():
    summary_ids = model.generate(
        **inputs,
        max_new_tokens=80,
        num_beams=4,
        do_sample=False,
        early_stopping=True
    )

summary = tokenizer.decode(summary_ids[0], skip_special_tokens=True)
print("Original paragraph:", paragraph)
print("\nSummary:", summary)
````

**Why is a decoder-only model suitable for a chatbot such as ChatGPT?**


A decoder-only model is suitable for a chatbot because it generates a response one token at a time using the previous conversation as context. Its causal attention lets each token use the tokens that came before it. This makes it useful for producing open-ended replies of different lengths. Instruction tuning helps the model follow requests and give more relevant answers.

### Question

Why do you sometimes get different outputs even when the prompt is the same?

**Hint:** Sampling introduces randomness when choosing the next token.


````python
sampling_prompt = "The future of artificial intelligence is"

# Leave the seed unset here to allow different samples on repeated runs.
for attempt in range(3):
    sampled_output = generator(
        sampling_prompt,
        max_new_tokens=40,
        do_sample=True,
        temperature=0.8,
        top_k=50,
        top_p=0.95,
        pad_token_id=generator.tokenizer.eos_token_id
    )
    print(f"Output {attempt + 1}: {sampled_output[0]['generated_text']}\n")
````

With sampling enabled, the model randomly selects the next token from a probability distribution instead of always choosing the most likely token. A different choice changes the context for the following tokens, so the same prompt can produce different continuations. Temperature, top-k, and top-p control how tokens are sampled. Resetting the seed to 42 before each generation makes the random choices reproducible under the same conditions; setting it once at the start does not make every call identical.
